```python
from pyspark.sql.functions import current_timestamp

base_path = "/Workspace/Users/astha.saraf93@gmail.com/databricks-genai-rag-agent-portfolio/data"

validation_path = f"{base_path}/validation_report.csv"
summary_path = f"{base_path}/summary_report.csv"
history_path = f"{base_path}/history.csv"

validation_df = (
    spark.read
    .option("header", True)
    .option("inferSchema", True)
    .csv(validation_path)
    .withColumn("loaded_at", current_timestamp())
)

summary_df = (
    spark.read
    .option("header", True)
    .option("inferSchema", True)
    .csv(summary_path)
    .withColumn("loaded_at", current_timestamp())
)

history_df = (
    spark.read
    .option("header", True)
    .option("inferSchema", True)
    .csv(history_path)
    .withColumn("loaded_at", current_timestamp())
)

validation_df.write.mode("overwrite").saveAsTable("workspace.default.bronze_validation_report")
summary_df.write.mode("overwrite").saveAsTable("workspace.default.bronze_summary_report")
history_df.write.mode("overwrite").saveAsTable("workspace.default.bronze_quality_history")

display(validation_df)
```
