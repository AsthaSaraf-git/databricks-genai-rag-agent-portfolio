```python
from pyspark.sql.functions import concat_ws, lit

quality_df = spark.table("workspace.default.gold_quality_by_table")

rag_docs = (
    quality_df
    .select(
        concat_ws(
            " ",
            lit("Table"),
            quality_df.table_name,
            lit("has quality score"),
            quality_df.quality_score.cast("string"),
            lit("with"),
            quality_df.failed_rules.cast("string"),
            lit("failed rules")
        ).alias("document")
    )
)

display(rag_docs)

rag_docs.write.mode("overwrite").saveAsTable(
    "workspace.default.rag_quality_documents"
)
```
